```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

!pip install dtale
!pip install dtale-app
import dtale
import dtale-app as dtale_app

dtale_app.USE_COLAB = True
dtale:show(data)

url = "https://raw.githubusercontent.com/ManarOmar/New-York-Airbnb-2019/master/AB_NYC_2019.csv"
data = pd.read_csv(url)

# Exploración básica
print(data.head())
print("Dimensiones:", data.shape)
data.info()
print("¿ID es único?:", data["id"].is_unique)
print("Total registros:", len(data))
print("Hosts únicos:", data["host_id"].nunique())
print(data.describe())

columnas = [
    "price", "minimum_nights", "number_of_reviews",
    "reviews_per_month", "calculated_host_listings_count", "availability_365"
]
print(data[columnas].describe())

# Imputación de nulos inicial en variable local
numericas = data[columnas].copy()
numericas = numericas.fillna(numericas.mean())

print("Nulos por columna:")
print(data.isnull().sum())

sin_resenas = data["number_of_reviews"] == 0
print("Anuncios sin reseñas:", sin_resenas.sum())

# Histograma general
data.hist(bins=10, figsize=(10, 6))
plt.tight_layout()
plt.show()

# Distribuciones categóricas
print(data["neighbourhood"].value_counts())
print(data["room_type"].value_counts())
print(data["neighbourhood_group"].value_counts())

media = data["reviews_per_month"].mean()
print("Media de reviews_per_month:", round(media, 2))

# Imputación de ceros para listings sin reseñas
data["reviews_per_month"] = data["reviews_per_month"].fillna(0)
print("Nulos restantes en reviews_per_month:", data["reviews_per_month"].isnull().sum())

numericas = data.select_dtypes(include=np.number)

# Boxplot del precio con outliers originales
data.boxplot(column="price", figsize=(8, 6))
plt.title("Boxplot Precios Originales")
plt.show()

# Definición de límites y filtrado
precio_minimo = 0
precio_maximo = 500  # Ajustable según el percentil o criterio de outliers deseado

datos_limpios = data[(data["price"] > precio_minimo) & (data["price"] <= precio_maximo)].copy()

print("Antes:", len(data), "anuncios")
print("Después:", len(datos_limpios), "anuncios")
print(datos_limpios["price"].describe())

# Boxplot tras limpiar outliers
datos_limpios.boxplot(column="price", figsize=(8, 6))
plt.ylabel("Precio por noche (USD)")
plt.title("Boxplot de precios filtrados")
plt.show()

# Histograma específico del precio
datos_limpios["price"].hist(bins=50, figsize=(10, 5))
plt.xlabel("Precio por noche (USD)")
plt.ylabel("Anuncios")
plt.title("Distribución del precio")
plt.show()

# Categorización del precio
corte_economico = 80
corte_medio = 200

def categoria_precio(precio):
    if precio <= corte_economico:
        return "economico"
    elif precio <= corte_medio:
        return "medio"
    else:
        return "premium"

datos_limpios["categoria"] = datos_limpios["price"].apply(categoria_precio)
print(datos_limpios["categoria"].value_counts())

datos_limpios.groupby("neighbourhood_group")["price"].agg(["count", "mean", "median"])
datos_limpios.groupby(["neighbourhood_group","room_type"])["price"].median().unstack()
datos_limpios.corr(numeric_only=True)
```
